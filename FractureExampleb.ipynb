```python
#some libraries 
import numpy as np
import math
import matplotlib.pyplot as plt
import pandas as pd
import random
import json
e = 0.03 #relaxation threshold 
k = 1.0 #spring constant representing elasticity, smaller is more elastic, larger is stiffer  
l = 1.5 #equilibrium distance, original is 2.0, so the smaller the more extreme the shrinking and crack opening and larger cells (try 1.7,1.3)
n = 100 # size of the array
fover = 1.1 #over relaxation factor
lattice = np.zeros([n,n,2]) #array with particle positions [0] = x, [1] = y
adjacent = np.zeros([n,n]).tolist() #array with list of neighbour particles
stress = np.zeros([n,n,3]) #sxx[0], syy[1], sxy[2] stress per particle 
```

```python
def triangular(): #create a triangular lattice, fill a square lattice(i,j) with particles and give them positions (not all cells are filled, empty cells remain zero, lattice is shifted by 2 so that no cells have zero positions)
    global lattice
    for i in range(n):
        for j in range(n):
            if (i%2==0 and j%2==0) or (i%2==1 and j%2==1): #fill cells where
                lattice[i][j][0] = i*math.sqrt(3)+2 
                lattice[i][j][1] = j+2
```

```python
def adjacent_particles():
    global adjacent
    neighbours = [(-1,-1),(-1,1),(0,-2),(0,2),(1,-1),(1,1)]
    for i in range(2,n-2):
        for j in range(2,n-2):
            if lattice[i][j].all() != 0:
                adjacent[i][j] = []
                for nb in neighbours:
                    if (i+nb[0]>-1 and i+nb[0]<n) and (j+nb[1]>-1 and j+nb[1]<n):
                        if lattice[i+nb[0]][j+nb[1]].all() !=0:
                            adjacent[i][j].append((i+nb[0],j+nb[1]))
```

```python
def force(i, j): 
    global stress, lattice #this will allow you to access and modify the variables declared outside this function
    xx, yy, sxx, syy, sxy, anc = 0, 0, 0, 0, 0, 0 
    
# distance between 2 particles 
    for nb in adjacent[i][j]:
        dx = lattice[i][j][0]-lattice[nb[0]][nb[1]][0]
        dy = lattice[i][j][1]-lattice[nb[0]][nb[1]][1]
        dd = math.sqrt(dx**2 + dy**2)
        if (dd != 0):
            uxi = dx/dd
            uyi = dy/dd
        else:  
            uxi = 0.0
            uyi = 0.0
        fn = -k*(dd-l)
        fx = fn*uxi
        fy = fn*uyi  
        
# overall force and stress
        xx = xx + fx
        yy = yy + fy
        sxx = sxx + fx*(l/2)*uxi
        syy = syy + fy*(l/2)*uyi
        sxy = sxy + fx*(l/2)*uyi
        anc = anc + k
    stress[i][j][0] = sxx
    stress[i][j][1] = syy
    stress[i][j][2] = sxy 
    
# over relaxation step          
    anc = 1.0/anc
    xx = xx * anc        
    yy = yy * anc        
    if (e < math.sqrt((xx*xx)+(yy*yy))): 
        lattice[i][j][0] = lattice[i][j][0] + xx * fover   
        lattice[i][j][1] = lattice[i][j][1] + yy * fover        
```

```python
def run_simulation():
    for z in range(2): #number of time steps
        for i in range(2,n-2): # leaving the fixed borders out
            for j in range(2,n-2):
                if lattice[i][j].all()!=0:
                    force(i, j)
```

```python
def fracturing(i,j):
    global mbreak, spring_to_break, spring_x, spring_y
    tb = 1
    ten_break = 0
    for nb in adjacent[i][j]:
        dx = lattice[i][j][0]-lattice[nb[0]][nb[1]][0]
        dy = lattice[i][j][1]-lattice[nb[0]][nb[1]][1]
        dd = math.sqrt(dx**2 + dy**2)
        fn = k*(dd-l)             
        if (fn >= 0.0):
            ten_break = (fn / tb) + random.randint(0,100)/100            
        if (ten_break > 1.0): 
            if (ten_break > mbreak):  
                mbreak = ten_break 
                spring_to_break = nb
                spring_x = i
                spring_y = j
```

```python
triangular()
adjacent_particles()
for h in range(5000):  
    mbreak = 0
    spring_to_break = 0
    spring_x, spring_y = 0, 0    
    run_simulation()    
    for i in range(2,n-2): 
        for j in range(2,n-2):
            if lattice[i][j].all()!=0:
                fracturing(i, j)                
    if (spring_to_break != 0):        
        for nb in adjacent[spring_x][spring_y]:
            if (adjacent[nb[0]][nb[1]] != 0.0): # if nb = 0.0 then its wall
                if nb == spring_to_break:
                    adjacent[nb[0]][nb[1]].remove((spring_x,spring_y))
                    adjacent[spring_x][spring_y].remove(nb)
    
    # make a plot every time step indicated in the brackets
    if (h%50==0):
        x = lattice[:, :, 0]
        x = x[x!=0]
        y = lattice[:, :, 1]
        y = y[y!=0]
        plt.axis('off')
        plt.rcParams["figure.figsize"] = (10,9)
        plt.scatter(x, y, s = 15)
        plt.show()
```
